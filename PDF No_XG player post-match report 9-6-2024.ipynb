```python
# ADD PLAYERS FORMATION
# READ FORMATIONS
formation = pd.read_excel(r"D:/Projects/football analysis/team foramion id (version 1).xlsx")

# ADD TEAMS FORMATIONS COLUMNS 
final_events['Team formation id']=final_events['Team formation id'].fillna(0)
final_events['Team formation id']=final_events['Team formation id'].astype('int64')
final_events['Involved']=final_events['Involved'].fillna('0')
final_events['Team player formation']=final_events['Team player formation'].fillna('0')

final_events['Team formation'] = pd.merge(final_events, formation, on='Team formation id', how='left')['formation'] 
final_events['Positions'] = pd.merge(final_events, formation, on='Team formation id', how='left')['positions']





# CREATE A TEMP DF TO HOLD TEAMS FORMATIONS AND ALL PLAYERS POSITIONS (combined_team_df)

# Initialize lists to store the dictionaries
home_team_data = []
away_team_data = []

# Filter the DataFrame to get the rows with 'Team set up'
team_setup_events = final_events[final_events['type/displayName'] == 'Team set up']

# Iterate through the filtered DataFrame
for i in range(0, len(team_setup_events), 2):
    # Home team dictionary
    home_team_dict = {
        "Involved": team_setup_events.iloc[i]["Involved"].split(', '),
        "Team player formation": team_setup_events.iloc[i]["Team player formation"].split(', '),
        "Positions": team_setup_events.iloc[i]["Positions"].split(', '),
        "Team formation": team_setup_events.iloc[i]["Team formation"],
        "Team id": team_setup_events.iloc[i]["teamId"]
    }
    home_team_data.append(home_team_dict)

    # Check if there is a corresponding away team row
    if i + 1 < len(team_setup_events):
        # Away team dictionary
        away_team_dict = {
            "Involved": team_setup_events.iloc[i + 1]["Involved"].split(', '),
            "Team player formation": team_setup_events.iloc[i + 1]["Team player formation"].split(', '),
            "Positions": team_setup_events.iloc[i + 1]["Positions"].split(', '),
            "Team formation": team_setup_events.iloc[i+1]["Team formation"],
            "Team id": team_setup_events.iloc[i+1]["teamId"]
        }
        away_team_data.append(away_team_dict)




def extract_positions(position_str):
    return position_str.split(',')

# Function to create player info dictionaries
def create_player_info(team_data):
    player_info = []
    for team in team_data:
        positions = extract_positions(team["Positions"][0])
        for involved, formation, position in zip(team["Involved"], team["Team player formation"], positions):
            player_info.append({
                "playerId": involved,
                "Team formation": team["Team formation"] ,
                "Team id":team['Team id'],
                "Team player formation": formation,
                "Position": position
                
            })
    return player_info

# Create player info for home team
home_team_players = create_player_info(home_team_data)
away_team_players = create_player_info(away_team_data)


# Convert list of dictionaries to DataFrame
home_team_df = pd.DataFrame(home_team_players)
away_team_df = pd.DataFrame(away_team_players)

# Display the DataFrame
# print("Home Team DataFrame")
# print(home_team_df)

# print("Away Team DataFrame")
# print(away_team_df)


combined_team_df = pd.concat([home_team_df, away_team_df], ignore_index=True)


# CREATE TEMP DF TO HOLD POSITIONS OF SUBBED ON PLAYERS (player_on_data)
player_on_data = final_events[final_events['type/displayName'] == 'Player on']
player_on_data = player_on_data.loc[:, [ 'playerId','teamId','Formation slot']]

player_on_data.rename(columns = {'Formation slot':'Team player formation','teamId':'Team id'}, inplace = True)

player_on_data = player_on_data.merge(combined_team_df[["Team player formation",'Team id', "Position"]],on = ["Team player formation", "Team id"], how ='left' )

player_on_data.rename(columns = {'Position':'New Position','Team player formation':'New formation'}, inplace = True)



# ADD SUBBED ON PLAYER POSITIONS TO ALL PLAYERS POSITIONS
combined_team_df =pd.merge(combined_team_df, player_on_data,  on=['playerId','Team id'], how='left')

combined_team_df["Team player formation"] = combined_team_df["New formation"].fillna(combined_team_df["Team player formation"])
combined_team_df["Position"] = combined_team_df["New Position"].fillna(combined_team_df["Position"])


combined_team_df.drop(columns=["New formation"], inplace=True)
combined_team_df.drop(columns=["New Position"], inplace=True)
combined_team_df.drop(columns=["Team id"], inplace=True)
combined_team_df.rename(columns = {'Team player formation':'Player formation'}, inplace = True)


final_events.drop(columns=["Positions","Team formation"], inplace=True)


# MERGE PLAYER POSITIONS TO FINAL_EVENTS
final_events= pd.merge(final_events, combined_team_df, on='playerId', how='left')
final_events
```

```python
#get it from network 'matchdetails matchid=' request
#Scrape shots from fotmob 

#############################################################################################################
match_id=4519290
#4330341
url = f'https://www.fotmob.com/api/matchDetails?matchId={match_id}'          ############################  CHANGE MATCH ID FROM FOTMOB###########################################
#############################################################################################################

response = requests.get(url)

if response.status_code == 200:
    data = response.text

    # Accessing the shotmap
    shotmap = data.split('matchId')
    #print(shotmap)



else:
    print("Failed to retrieve data. Status Code:", response.status_code)


# #get shot mop
# shot_map=pd.DataFrame(shotmap)
# shot_map=shot_map.rename({'id':'fotmob_id'}, axis=1)
# shot_map
# shot_map.to_excel('D:/Projects/football analysis/23-24/Wessam abo ali/Shots.xlsx',index=False)

f=response.text.split('shotmap' + '":{"shots":[')[1]
d=f.split(',"lineup":')[0]  #,"lineup":
dd=d.split('],"Periods":')              #],"Periods":
data=dd[0]

data='['+data+']'
data=json.loads(data)
shots=pd.DataFrame(data)
shots_data_source="shots.xlsx"

shots.to_excel(shots_data_source)


r=response.text.split('{"general":')[1]
r=r.split(',"header":')[0]  
r=json.loads(r)
r=pd.DataFrame(r)



r['home_team_name']=r.loc['name']['homeTeam']
r['home_team_id']=r.loc['id']['homeTeam']
r['away_team_name']=r.loc['name']['awayTeam']
r['away_team_id']=r.loc['id']['awayTeam']
r.iloc[0]

fotmob_match_details = pd.DataFrame(r.iloc[0]).transpose()


fotmob_match_details[['homeColor', 'awayColor']] = fotmob_match_details['teamColors'].apply(lambda x: pd.Series([x['home'], x['away']]))
#list(fotmob_match_details.columns)

competition_name=fotmob_match_details['leagueName']['darkMode']
round=fotmob_match_details['leagueRoundName']['darkMode']
competition_id=fotmob_match_details['parentLeagueId']['darkMode']


def save_image_from_url(image_url, save_path, background_color=(38, 38, 38)):
    try:
        # Make a GET request to the image URL
        response = requests.get(image_url)

        # Check if the request was successful (status code 200)
        if response.status_code == 200:
            # Open the image using PIL
            image = Image.open(BytesIO(response.content))

            # Convert image to RGBA if it has transparency
            if image.mode in ('RGBA', 'LA') or (image.mode == 'P' and 'transparency' in image.info):
                image = image.convert('RGBA')

            # Create a new image with the specified background color
            new_image = Image.new("RGBA", image.size, background_color)
            new_image.paste(image, (0, 0), image)

            # Save the new image to the specified path as PNG
            new_image.save(save_path, format='PNG')

            print(f"Image saved successfully at: {save_path}")
        else:
            print(f"Failed to retrieve image. Status code: {response.status_code}")
    except Exception as e:
        print(f"Error: {e}")

competition_image_url = f"https://images.fotmob.com/image_resources/logo/leaguelogo/dark/{competition_id}.png" ##### Replace with your FOTMOB COMPETTION image URL#################333
save_path = "competition_image.png"   
save_image_from_url(competition_image_url, save_path)
```

```python
def images_to_pdf(image_paths, output_pdf, default_page_width, default_page_height, bg_color=(38, 38, 38)):
    temp_files = []  # Keep track of temporary files
    
    # Images that need original height
    special_images = ['final_3rd_passes.png', 'shot_map.png']
    
    for i, image_path in enumerate(image_paths):
        print(f"Processing image {i+1}/{len(image_paths)}: {image_path}")
        
        if not os.path.exists(image_path):
            print(f"Warning: Image not found - {image_path}")
            continue
            
        img = Image.open(image_path)
        
        # Check if this is one of the special images
        image_name = os.path.basename(image_path)
        if image_name in special_images:
            # Use original height for special images
            page_width = default_page_width
            page_height = img.size[1] * 0.264583  # Convert original height to mm
            # Create new PDF instance for this page with custom dimensions
            pdf = FPDF(unit="mm", format=(page_width, page_height))
        else:
            # Use default dimensions for other images
            page_width = default_page_width
            page_height = default_page_height
            # Create new PDF instance for this page with default dimensions
            pdf = FPDF(unit="mm", format=(default_page_width, default_page_height))
        
        # Convert image size from pixels to mm for PDF
        img_width, img_height = img.size
        pdf_width_px = int(page_width / 0.264583)
        pdf_height_px = int(page_height / 0.264583)
        
        # Create a background canvas with the specified color and PDF page size
        bg_image = Image.new("RGB", (pdf_width_px, pdf_height_px), bg_color)
        
        # Calculate position to center the image on the background
        offset_x = (pdf_width_px - img_width) // 2
        offset_y = (pdf_height_px - img_height) // 2
        
        # Paste the image onto the center of the background canvas
        bg_image.paste(img, (offset_x, offset_y))
        
        # Create unique temporary file for each image
        temp_path = f"temp_image_{i}.jpg"
        temp_files.append(temp_path)
        bg_image.save(temp_path, "JPEG")
        
        # Add page and image
        pdf.add_page()
        pdf.image(temp_path, x=0, y=0, w=page_width, h=page_height)
        
        # Save individual page
        page_pdf = f"page_{i}.pdf"
        pdf.output(page_pdf)
        
        # Close the images to free up memory
        img.close()
        bg_image.close()
    
    # Merge all PDFs
    from PyPDF2 import PdfMerger
    merger = PdfMerger()
    
    # Add all pages to merger
    for i in range(len(image_paths)):
        page_pdf = f"page_{i}.pdf"
        if os.path.exists(page_pdf):
            merger.append(page_pdf)
    
    # Write final PDF
    try:
        merger.write(output_pdf)
        merger.close()
        print(f"PDF saved as {output_pdf}")
    except Exception as e:
        print(f"Error: {e}")
    finally:
        # Clean up temporary files
        for temp_file in temp_files:
            if os.path.exists(temp_file):
                os.remove(temp_file)
        # Clean up individual page PDFs
        for i in range(len(image_paths)):
            page_pdf = f"page_{i}.pdf"
            if os.path.exists(page_pdf):
                os.remove(page_pdf)

# Define the folder and list of specific images
image_folder = folder_path
specific_images = [
    'passing_network.png',
    'pass_end_zones.png',
    'home_final_3rd_passes.png',
    'away_final_3rd_passes.png',
    'final_3rd_entries.png',
    'box_entries.png',
    'gk_passes.png',
    'final_3rd_rec.png',
    'def_actions_zones.png',
    'home_shot_map.png',
    'away_shot_map.png'
]

# Print all files in the folder for debugging
print("Files in folder:")
print(os.listdir(image_folder))

# Gather images, using case-insensitive comparison
specific_image_paths = []
for img_name in specific_images:
    matching_files = [f for f in os.listdir(image_folder) 
                     if f.lower() == img_name.lower()]
    if matching_files:
        specific_image_paths.append(os.path.join(image_folder, matching_files[0]))
    else:
        print(f"Warning: {img_name} not found in folder")

# Gather "vs" images with their creation dates
vs_image_paths_with_dates = [
    (os.path.join(image_folder, img), os.path.getctime(os.path.join(image_folder, img)))
    for img in os.listdir(image_folder)
    if 'vs' in img.lower() and img.lower().endswith(('.png', '.jpg', '.jpeg'))
]

# Sort by creation date (oldest first)
vs_image_paths_with_dates.sort(key=lambda x: x[1])

# Extract sorted image paths
vs_image_paths = [path for path, _ in vs_image_paths_with_dates]

# Print found images for debugging
print("\nVS images sorted by date:")
print(vs_image_paths)

# Print found images for debugging
print("\nSpecific images found:")
print(specific_image_paths)


# Combine specific images first, then "vs" images
all_images = specific_image_paths + vs_image_paths
# Get dimensions of the first "vs" image
if vs_image_paths:
    with Image.open(vs_image_paths[0]) as vs_img:
        page_width, page_height = vs_img.size[0] * 0.264583, vs_img.size[1] * 0.264583  # Convert pixels to mm
else:
    page_width, page_height = 210, 297  # Default to A4 if no "vs" image found

# Generate PDF with background color "grey 15" (RGB 38, 38, 38)
images_to_pdf(
    all_images, 
    f"{teams_id_name[0]['team1_name']} vs {teams_id_name[0]['team2_name']} - {date.today()}.pdf",
    page_width, 
    page_height, 
    bg_color=(38, 38, 38)
)
```
